У Вас є великий DataFrame df, що представляє транзакції продажу в магазині. Цей DataFrame містить понад 1 мільйон рядків і кілька стовпців, зокрема «product_name», «product_category», «product_price» і «transaction_date».

І «product_name», і «product_category» є рядками, що представляють численні, але обмежені унікальні назви продуктів і категорії відповідно.

Напишіть код для ефективного перетворення стовпців «product_name» і «product_category» на категорійний тип для ефективного використання пам’яті.
Спробуйте вирішити це завдання, мінімізуючи використання пам'яті та зберігаючи читабельність коду. Крім того, виміряйте використання пам’яті до та після перетворення, щоб продемонструвати підвищення ефективності.

Порівняйте кількість памʼяті, яку займає наданий набір даних і новий, створений Вами. В лекції був приклад. В скільки разів менше памʼяті займає новий датафрейм?

````python
import pandas as pd
import numpy as np

np.random.seed(0)

products = ['apple', 'banana', 'cherry', 'dates', 'elderberry']
categories = ['fruit', 'fruit', 'fruit', 'fruit', 'fruit']

N_ROWS = 10**6
df = pd.DataFrame({
    'product_name': np.random.choice(products, N_ROWS),
    'product_category': np.random.choice(categories, N_ROWS),
    'product_price': np.random.uniform(1, 10, N_ROWS),
    'transaction_date': pd.date_range(start='1/1/2000', periods=N_ROWS, freq='h')
})
````

````python
memory_before = df.memory_usage(index=False).sum()

categorical_columns = ['product_name', 'product_category']

df[categorical_columns] = df[categorical_columns].astype('category')

memory_after = df.memory_usage(index=False, deep=True).sum()

reduction_percent = (1 - memory_after / memory_before) * 100
times_less = memory_before / memory_after

print(f"Memory before: {memory_before / 1024 ** 2:.2f} MB")
print(f"Memory after: {memory_after / 1024 ** 2:.2f} MB")
print(f"Memory reduction: {reduction_percent:.2f}%")
print(f"Після оптимізації DataFrame використовує у {times_less:.2f} рази менше памʼяті.")
````

У відредагованій версії:

- виміряла використання памʼяті до перетворення за допомогою memory_usage(deep=True);
- створила список categorical_columns тільки з колонками product_name та product_category;
- перетворила ці колонки на тип category, оскільки вони містять обмежену кількість повторюваних текстових значень;
- не змінювала product_price та transaction_date, бо це числова колонка і колонка з датою;
- повторно виміряла використання памʼяті після перетворення;
- порахувала, на скільки % зменшилося використання памʼяті та у скільки разів оптимізований DataFrame займає менше памʼяті.

````python
def calculate_variance(numbers):
    mean = sum(numbers) / len(numbers)

    squared_deviations = [
        (number - mean) ** 2
        for number in numbers
    ]

    variance = sum(squared_deviations) / len(numbers)

    return variance


input_list = [75, 80, 85, 90, 95]

result = calculate_variance(input_list)

print(int(result))
````
