6. Трансформуйте колонки 'Ваш вік', 'Зарплата у $$$ за місяць, лише ставка після сплати податків' у тип даних float. УВАГА! Перед трансформацією подивіться, які дані містять ці колонки. Аби успішно трансформувати тип даних, треба спочатку привести рядки до того виду, який буде підходити, аби потім рядок зробити типом float.

Пропоную наступний алгоритм дій:
    1. Подивитись, які дані містяться в цій колонці. Виведіть перші і останні 5 записів. Чи однаковий формат цих даних? (спойлер: ні, бо інакше все було б просто :))
    2. Нам треба враховуючи різний формат даних перетворити рядок у колонці `Timestamp` на тип `datetime64[ns]`. Як це зробити? Для перетворення рядка на `datetime64` пропоную скористатись `datetime.datetime.strptime`.
    Для цього треба імпортувати бібліотеку `datetime`. Я пропоную написати функцію для обробки кожного окремого рядка, яка враховує довжину рядка і на основі цього застосовує різний формат даних дати.
    Як прописувати формат даних можна переглянути [тут](https://www.w3schools.com/python/python_datetime.asp).
    3. Коли функція для обробки кожного окремого рядка готова, застосуйте її для всіх даних з допомогою методу `apply`.

```python
from datetime import datetime

print(df['Timestamp'].head())
print()
print(df['Timestamp'].tail())
```

```python
def parse_timestamp(x):
    x = str(x).strip()
    
    if len(x) == 19:
        return datetime.strptime(x, '%m.%d.%Y %H:%M:%S')
    elif len(x) == 16:
        return datetime.strptime(x, '%m.%d.%Y %H:%M')
    else:
        return None
```

```python
df['Timestamp'] = df['Timestamp'].apply(parse_timestamp).astype('datetime64[ns]')
df['Timestamp']
```

9. Для колонок типу даних object виведіть топ 10 найбільш частих значень з кількістю рядків, в яких це значення зустрічається. Зробіть це для кожної колонки окремо. Наприклад, в циклі.  
Що цікавого бачите? Які висновки можна зробити? Що особисто Вас вразило?

```python
for col in object_cols:
    counts = df[col].value_counts(dropna=False).head(10)
    percents = (df[col].value_counts(dropna=False, normalize=True).head(10) * 100).round(2).astype(str) + '%'
    
    result = pd.DataFrame({
        'count': counts,
        'percent': percents
    })
    print(f'\n{"="*60}')
    print(f'Колонка: {col}')
    print(result)
```

10. Виведіть медіанну (медіана - це значення, менше якого є рівно 50% вибірки і більше якого є теж рівно 50% вибірки)

11. Є така штука як сортування значень категоріальної змінної. Нам це зараз стане в нагоді, адже рівні англійської можна посортувати.
Давайте відсотруємо рівні англ. і тайтли. Просто запустіть код нижче і проаналізуйте, що він робить.

```python
df['Знання англійської мови'] = pd.Categorical(df['Знання англійської мови'],
                      categories=['Elementary','Pre-Intermediate','Intermediate','Upper-Intermediate','Advanced'],
                      ordered=True)

df['Ваш тайтл'] = pd.Categorical(df['Ваш тайтл'],
                      categories=['Intern/Trainee', 'Junior', 'Немає тайтлу', 'Middle', 'Senior', 'Tech Lead', 'Team Lead', 'Architect', 'Head', 'Manager'],
                      ordered=True)
```

У цьому кроці текстові значення перетворили на впорядковані категорії.  
Завдяки цьому рівні англійської і тайтли можна сортувати у правильному порядку, а не за алфавітом.

12. Зробіть зводну таблицю, де стовпці - це 'Ваш тайтл', рядки - 'Знання англійської мови', а значення - медіанна ЗП. Які висновки можна зробити?
Тут ще для зручності можна зробити градієнтне забарвлення і відформатувати значення для легшого перегляду.

```python
pivoted = df.pivot_table(index='Ваш тайтл',
                         columns='Знання англійської мови',
                         values='Зарплата у $$$ за місяць, лише ставка після сплати податків',
                         aggfunc='median',
                         observed=False
                        ).round(0)

pivoted.style.background_gradient(cmap='YlGn').format('{:.0f}')
```

**Висновки**

- Чим вищий рівень англійської, тим зазвичай вища зарплата.
- Чим вищий тайтл, тим теж вища медіанна зарплата.
- Найменші зарплати у Intern і Junior, найбільші — у Senior, Lead, Architect, Head.
- У деяких групах є пропуски, тому там даних мало або немає зовсім.
- Є і винятки, тому залежність не всюди абсолютно однакова. Бачимо Tech Lead при рівні Elementary має середню зп даної посади.

13. Виведіть результат методу describe() по колонці розміру ЗП для топ 5 за частотою зустрічання в даних областей (В якій області ви зараз живете?). Зробіть висновки, де вигідніше жити з точки зору ЗП і чи суттєва різниця на Ваш погляд для прийняття рішення про переїзд?

```python
top_5_regions = df['В якій області ви зараз живете?'].value_counts().head(5).index

salary_by_region = (
    df[df['В якій області ви зараз живете?'].isin(top_5_regions)]
    .groupby('В якій області ви зараз живете?')['Зарплата у $$$ за місяць, лише ставка після сплати податків']
    .describe().round(1)
    .sort_values(by='50%', ascending=False)
)

print(salary_by_region)
```

- Найвища медіанна зарплата серед топ-5 регіонів — у Києві та області.
- Львів має дуже близькі до Києва показники за середньою та медіанною зп, тому суттєвої різниці між ними майже немає.
- Далі - Вінниця, а в Івано-Франківську та Дніпрі медіанна зарплата нижча.
- Різниця між регіонами є, але для рішення про переїзд варто враховувати не лише зарплату, а й вартість життя та інші умови.
- Мах значення великі, але орієнтуватися краще на медіану, бо вона краще показує типову зарплату.

14. Як зростає зарплата з роками? Для цього нам треба відформатувати колонку "Загальний стаж роботи за спеціальністю" аналогічно до того, як ми це робили для колонки 'Знання англійської мови' і потім вивести медіанні значення для кількості років стажу.

```python
df['Загальний стаж роботи за спеціальністю'] = pd.Categorical(
    df['Загальний стаж роботи за спеціальністю'],
    categories=[
        'Менше як 3 місяці', '3 місяці', 'Пів року', '1 рік', '1,5 року', '2 роки', '3 роки', '4 роки', '5 років', '6 років',
        '7 років', '8 років', '9 років', '10 років', '11 років', '12 років', '13 років', '14 років', '15 і більше років'],
    ordered=True
)
```

```python
work_experience = (
    df.groupby('Загальний стаж роботи за спеціальністю')['Зарплата у $$$ за місяць, лише ставка після сплати податків']
    .median() 
)

print(work_experience)
```

**Висновки**

- Чим більший стаж, тим вища зарплата 
- На початку є невеликі коливання, але загальна тенденція очевидна.
- Найменші зарплати у тих, хто має мало досвіду, найбільші — у спеціалістів з великим стажем.
- Після 8–10 років зарплата вже не росте так рівно, але все одно залишається високою.
- Бачимо відхилення, наприклад із стажем 8 років отримує більше, ніж 9-10, тому тут не тільки про стаж, але і навички людини залежить.
